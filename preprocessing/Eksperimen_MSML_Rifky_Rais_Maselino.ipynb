# **4. Exploratory Data Analysis (EDA)**

Pada tahap ini, Anda akan melakukan **Exploratory Data Analysis (EDA)** untuk memahami karakteristik dataset.

Tujuan dari EDA adalah untuk memperoleh wawasan awal yang mendalam mengenai data dan menentukan langkah selanjutnya dalam analisis atau pemodelan.

```python
# EDA for Numerical Columns
numerical = df.select_dtypes(include=['int64', 'float64'])
categorical = df.select_dtypes(include=['object'])
correlation = numerical.corr()

plt.figure(figsize=(10,6))
correlation_matrix = numerical.corr()
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt='.2f', linewidth=0.5)
plt.title('Correlation Matrix')
plt.show()

plt.figure(figsize=(10,6))
sns.pairplot(numerical)
plt.show()
```

```python
plt.figure(figsize=(12, 6))
for i, column in enumerate(numerical.columns, 1):
  plt.subplot(3, 5, i)
  sns.histplot(df[column], bins=30, kde=True, color='blue')
  plt.title(f'Histogram of {column}')

plt.tight_layout()
plt.show()
```

```python
# EDA for Categorical Columns and Outlier Detection

plt.figure(figsize=(14,10))
for i, column in enumerate(categorical.columns, 1):
  plt.subplot(3, 4, i)
  sns.countplot(data=df[column], palette='viridis')
  plt.title(f"Distribusi {column}")

plt.tight_layout()
plt.show()

plt.figure(figsize=(14,10))
for i, column in enumerate(numerical.columns, 1):
  plt.subplot(3, 5, i)
  sns.boxplot(data=df[column], palette='viridis')
  plt.title(f"Distribusi {column}")

plt.tight_layout()
plt.show()
```

```python
# Missing Values Detection
df.isnull().sum()
```

```python
# Duplicated Values Detection
df.duplicated().sum()
```

```python
# Handling Missing and Duplicated Values
df.dropna(inplace=True)
df.drop_duplicates(inplace=True)
```

```python
# Outlier Detection and Handling
for col in ["Calories_Burned", "BMI"]:
    q1 = df[col].quantile(0.25)
    q3 = df[col].quantile(0.75)
    iqr = q3 - q1

    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr

    df[col] = df[col].clip(lower, upper)
```

```python
# Encoding Categorical Columns
df["Gender"] = LabelEncoder().fit_transform(df["Gender"])
df["Workout_Type"] = LabelEncoder().fit_transform(df["Workout_Type"])
```

```python
# Saving Preprocessed Data
df.to_csv('gym_dataset_preprocessed.csv', index=False)

print("Dataset baru berhasil disimpan dengan nama: gym_dataset_preprocessed.csv")
print(f"Jumlah Fitur: {df.shape[1]}")
print(df.head())
```
